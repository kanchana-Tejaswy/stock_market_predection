```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.figure(figsize=(15,5))

google_data['Close'].plot()

plt.xlabel("Years")
plt.ylabel("Closing Price")
plt.title("Closing Price of Google Data")
plt.show()
```

```python
def plot_graph(figsize, values, column_name):
    plt.figure(figsize=figsize)
    
    values.plot()
    
    plt.xlabel("Years")
    plt.ylabel(column_name)
    plt.title(f"{column_name} of Google Data")
    
    plt.show()
```

```python
google_data.columns
```

```python
for column in google_data.columns:
    plot_graph((15,5),google_data[column], column)
```

```python
google_data['MA_for_250_days'] = google_data['Close'].rolling(window=250).mean()
```

```python
google_data[['Close', 'MA_for_250_days']].tail()
```

```python
plot_graph((15,5), google_data['MA_for_250_days'], 'MA_for_250_days')
```

```python
plot_graph(
    (15,5),
    google_data[['Close', 'MA_for_250_days']],
    'MA_for_250_days'
)
```

```python
google_data['MA_for_100_days'] = google_data['Close'].rolling(window=100).mean()

plot_graph(
    (15,5),
    google_data[['Close', 'MA_for_100_days']],
    'MA_for_100_days'
)
```

```python
plot_graph(
    (15,5),
    google_data[['Close', 'MA_for_100_days', 'MA_for_250_days']],
    'Moving Averages'
)
```

```python
google_data['percentage_change_cp'] = google_data['Close'].pct_change()

google_data[['Close', 'percentage_change_cp']].head()
```

```python
plot_graph((15,5), google_data['percentage_change_cp'],'percentage_change')
```
